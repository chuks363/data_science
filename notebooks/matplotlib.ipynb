## 🔹 6. Heatmaps

### ✅ Heatmap Example

```python
import matplotlib.pyplot as plt
import numpy as np

# Simulate temperature data for 7 days over 10 weeks
np.random.seed(0)
temperatures = np.random.uniform(low=20, high=35, size=(7, 10))

plt.imshow(temperatures, cmap='hot', interpolation='nearest')
plt.colorbar(label="Temperature (°C)")
# Label axes
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
plt.yticks(ticks=range(7), labels=days)
plt.xticks(ticks=range(10), labels=[f"Week {i+1}" for i in range(10)], rotation=45)

plt.title("Average Daily Temperatures Over 10 Weeks")
plt.xlabel("Week")
plt.ylabel("Day")
plt.tight_layout()
plt.show()

```

````python
import matplotlib.pyplot as plt
import numpy as np

# Simulate temperature data for 7 days over 10 weeks
np.random.seed(0)
temperatures = np.random.uniform(low=20, high=35, size=(7, 10))

plt.imshow(temperatures, cmap='hot', interpolation='nearest')
plt.colorbar(label="Temperature (°C)")
# Label axes
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
plt.yticks(ticks=range(7), labels=days)
plt.xticks(ticks=range(10), labels=[f"Week {i+1}" for i in range(10)], rotation=45)

plt.title("Average Daily Temperatures Over 10 Weeks")
plt.xlabel("Week")
plt.ylabel("Day")
plt.tight_layout()
plt.show()
````

## 🔹 9. Subplots (Multiple Plots)

### ✅ Using `subplot()`

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
product_A_sales = [150, 200, 250, 220, 300]
product_B_sales = [100, 180, 230, 210, 280]

plt.figure(figsize=(10, 4))
# First subplot: Product A
plt.subplot(1, 2, 1)
plt.plot(months, product_A_sales, marker='o', color='blue')
plt.title("Product A Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
# Second subplot: Product B
plt.subplot(1, 2, 2)
plt.plot(months, product_B_sales, marker='s', color='green')
plt.title("Product B Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.tight_layout()
plt.show()
```

````python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
product_A_sales = [150, 200, 250, 220, 300]
product_B_sales = [100, 180, 230, 210, 280]

plt.figure(figsize=(10, 4))
# First subplot: Product A
plt.subplot(1, 2, 1)
plt.plot(months, product_A_sales, marker='o', color='blue')
plt.title("Product A Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
# Second subplot: Product B
plt.subplot(1, 2, 2)
plt.plot(months, product_B_sales, marker='s', color='green')
plt.title("Product B Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    'Year': [2018, 2019, 2020, 2021],
    'Sales A': [200, 220, 250, 300],
    'Sales B': [180, 210, 240, 270]
})

df.plot(x='Year', y=['Sales A', 'Sales B'], kind='line', marker='o')
plt.title("Yearly Sales Comparison")
plt.ylabel("Sales")
plt.grid(True)
plt.savefig("my_plot.png", dpi=300)
plt.show()
````

````python
import matplotlib.pyplot as plt
import random
from datetime import datetime, timedelta

# --- Generate mock timestamps (3-hour intervals over 5 days) ---
timestamps = [datetime.now() + timedelta(hours=3*i) for i in range(40)]

# --- Generate mock weather data ---
temperatures = [random.uniform(15, 30) for _ in timestamps]       # °C
humidities = [random.randint(40, 90) for _ in timestamps]         # %
wind_speeds = [random.uniform(1, 10) for _ in timestamps]         # m/s

# --- Plot the mock data ---
fig, axs = plt.subplots(3, 1, figsize=(12, 10), sharex=True)

# Temperature
axs[0].plot(timestamps, temperatures, color='tomato', marker='o')
axs[0].set_title('Temperature Trend (°C)')
axs[0].set_ylabel('Temp (°C)')
axs[0].grid(True)

# Humidity
axs[1].plot(timestamps, humidities, color='skyblue', marker='s')
axs[1].set_title('Humidity Trend (%)')
axs[1].set_ylabel('Humidity (%)')
axs[1].grid(True)

# Wind Speed
axs[2].plot(timestamps, wind_speeds, color='seagreen', marker='^')
axs[2].set_title('Wind Speed Trend (m/s)')
axs[2].set_ylabel('Wind Speed (m/s)')
axs[2].set_xlabel('Date & Time')
axs[2].grid(True)

# Format x-axis
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
````

*Mini challenge: Weather Visualization*

Use an API to fetch weather data and plot temperature trends, wind speeds, and humidity with subplots.


````python
import requests
import matplotlib.pyplot as plt
from datetime import datetime

# --- Step 1: User Inputs ---
API_KEY = input("Enter your OpenWeatherMap API key: ").strip()
CITY = input("Enter city name (e.g., London): ").strip()
URL = f'https://api.openweathermap.org/data/2.5/forecast?q={CITY}&appid={API_KEY}&units=metric'

# --- Step 2: Fetch Weather Data ---
response = requests.get(URL)
data = response.json()

# --- Step 3: Check for API Errors ---
if 'list' not in data:
    print("❌ Error fetching data:", data.get("message", "Unknown error"))
    exit()

# --- Step 4: Extract Forecast Info ---
timestamps = []
temperatures = []
humidities = []
wind_speeds = []

for entry in data['list']:
    timestamps.append(datetime.strptime(entry['dt_txt'], '%Y-%m-%d %H:%M:%S'))
    temperatures.append(entry['main']['temp'])
    humidities.append(entry['main']['humidity'])
    wind_speeds.append(entry['wind']['speed'])

# --- Step 5: Plot Subplots ---
fig, axs = plt.subplots(3, 1, figsize=(12, 10), sharex=True)

axs[0].plot(timestamps, temperatures, color='orangered', marker='o')
axs[0].set_title(f'Temperature Trend in {CITY.title()}')
axs[0].set_ylabel('°C')
axs[0].grid(True)

axs[1].plot(timestamps, humidities, color='dodgerblue', marker='s')
axs[1].set_title('Humidity Trend')
axs[1].set_ylabel('%')
axs[1].grid(True)

axs[2].plot(timestamps, wind_speeds, color='seagreen', marker='^')
axs[2].set_title('Wind Speed Trend')
axs[2].set_ylabel('m/s')
axs[2].set_xlabel('Date & Time')
axs[2].grid(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````
